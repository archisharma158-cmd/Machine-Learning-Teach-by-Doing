```python
positive_points = np.array([(-2,3), (0,1), (2,-1), (6,2)])
negative_points = np.array([(-2,1), (0,-1), (2,-3), (-2,-1)])
```

```python
#Combining the data points
x = np.vstack((positive_points, negative_points))
y = np.array([1]*len(positive_points) + [-1]*len(negative_points))
```

```python
#Perceptron Alogirthm
def perceptron(x,y, steps=10):
    theta = np.zeros(x.shape[1])
    theta0 = 0
    for m in range(steps):
        for n, xn in enumerate(x):
            if y[n] * (np.dot(xn, theta) + theta0) <= 0:
                theta += y[n] * xn
                theta0 += y[n]
                break
        plot_decision_boundary(theta, theta0, m+1)
    return theta, theta0
```

```python
def plot_decision_boundary(theta, theta0, iteration) :
    plt.figure(figsize=(8,6))
    plt.scatter(positive_points[:,0], positive_points[:,1], color='orange', label='Positive class')
    plt.scatter(negative_points[:,0], negative_points[:,1], color='green', label='Negative class')

    x_value = np.linspace(-3,3,2)

    if theta[1] != 0:
        y_values = -(theta[0] * x_values + theta0) / theta[1]
        plt.plot(x_values, y_values, label=f'Iteration{iteration}')
    else:
        x_val = -theta0 / theta[0]
        plt.axvline(x=x_val, label=f'Iteration{iteration}')


    plt.xlim(-3,3)
    plt.ylim(-4,4)
    plt.xlabel('x1')
    plt.ylabel('x2')
    plt.title(f'Desicion Boundary at Iteration {iteration}')
    plt.legend()
    plt.grid(True)
    plt.show()
```
